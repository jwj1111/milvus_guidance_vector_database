# pymilvus准备了方便的数据接口与用户管理操作

```python
from pymilvus import connections, utility
```

```python
# 连接数据库
connections.connect(uri="http://localhost:19530", token="root:Milvus")
# connections.connect(host="127.0.0.1", port="19530", token="root:Milvus")
```

```python
# 更新密码
utility.update_password(user_name="root", old_password="Milvus", new_password="www")
```

```python
# 创建新用户
utility.create_user(user="aaa", password="www")
```
